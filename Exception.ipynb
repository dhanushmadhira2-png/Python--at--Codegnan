```python
# Exception Handling Block
try:
    a = int(input())
    b = int(input())
    c = a+b
    print(c)
except:
    print("Something wrong in try block")
else:
    print("There is no exception")
finally:
    print("Program done")
```

```python
# Exception Handling Block
#  Exeception with exception class
try:
    a = 10
    b = 0
    c = a/b
    print(c)
except Exception as e:
    print(f"Something wrong in try block:{e}")
   
```

```python
# Exception Handling Block
#  Define multiple exceptions with difference except block
try:
    a = 10
    b = 0
    c = a/b
    print(c)
except ZeroDivisionError as e:
    print(f" Zero division not possible: {e}")
except IndexError as e:
    print(f"We can't access element at this position:{e}")
except Exception as e:
    print(f"Something wrong in try block:{e}")
```

```python
# Exception Handling Block
#  Define multiple exceptions with difference except block
try:
    a = [1,2,3,4,5]
    c = a[10]
    print(c)
except ZeroDivisionError as e:
    print(f" Zero division not possible: {e}")
except IndexError as e:
    print(f"We can't access element at this position:{e}")
except Exception as e:
    print(f"Something wrong in try block:{e}")
```
