```python
# Fibonacci number
def fib(n):
    if n==0 or n==1:
        return 1
    return fib(n-1)+fib(n-2)
fib(3)
```

```python
# sum of natural number
def sum(n):
    if n==1:
        return 1
    return n+sum(n-1)
sum(5)
```

```python
# Product of natural numbers
def pro(n):
    if n==1:
        return 1
    return n*pro(n-1)
pro(5)
```

```python
# Find the Binary number
def binary(n):
    if n==0:
        return "0"
    if n==1:
        return "1"
    return binary(n//2)+str(n%2)
binary(5)
```

```python
# sum of two numbers using lambda
add = lambda x, y: x + y
print(add(2,5))
```

```python
# find square of all list of integer using lambda and map()
arr = [1,2,3,4,5]
res = list(map(lambda x : x**2, arr))
print(res)
```

```python
# find even numbers in list using lambda and filter()
arr = [1,2,3,4,5]
res = list(filter(lambda x : x%2==0, arr))
print(res)
```

```python
# single line if-else syntax
# True_exp if condirtion else False_exp
#find the weather the nummber is even or odd using lamda
n = int(input())
even_or_odd = lambda x:"odd" if x%2 else "even"
print(even_or_odd(n))
```
