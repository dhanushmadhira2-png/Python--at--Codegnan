```python
##Output Formatting
#1. Old style Formatting (using comma,)
name = "dhanu"
age = 20
marks = [85,98,65,85,78,98]
#My name dhanu and i am a 20 year old
print("My name ", name ," and i am a ",str(age)," years old")
```

```python
## By using access specifiers
# integer - %d
# float - %f
# string - %s
name = "Ravi"
age = 20
marks = [85,98,65,85,78,98]
#My name Ravi and i am a 20 year old
print("My name %s and i am a %d years old"%(name,age))
```

```python
## output formatting
# 3 Dot Format
name = "Ravi"
age = 20
marks = [85,98,65,85,78,98]
#My name Ravi and i am a 20 year old
print("My name {} and i am a {} years old".format(name,age))
print("My name {} and i am a {} years old".format(age,name))
```

```python
##4.f-Format
name = "Ravi"
age = 20
marks = [85,98,65,85,78,98]
#My name Ravi and i am a 20 year old
print(f"My name {name} and i am a {age} years old")
percentage = 98.2568
print(f"My name {name} and i am a {age} years old and my percentage is {percentage}")
print(f"My name {name} and i am a {age} years old and my percentage is {percentage:.2f}")
```
