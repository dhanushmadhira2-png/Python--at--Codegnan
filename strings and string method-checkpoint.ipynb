```python
#### String Methods
- Case Conversion Methods
   - upper(): It converts entire string into uppercase
   - lower(): It converts entire string into uppercase
   - title(): It converts every first character into uppercase and remaining into lowercase 
   - swapcase(): It converts lower into upper and upper to lowercase
   - capitalize(): It converts first character of string into uppercase and remaining into lowercase
```

```python
string = "The ocean is blue and sky is BLUE"
lower = string.lower()
print(lower)
print(string.upper())
print(string.title())
print(string.swapcase())
print(string.capitalize())
```

```python
# white space removig methods
# strip(): It removes the white space to both sides of string
# lstrip(): It removes the leading white space
# rstrip(): It removes the lagging white space
string ="         Codegnan           "
print(string)
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

```python
# white space removig methods
# strip(): It removes the white space to both sides of string
# lstrip(): It removes the leading white space
# rstrip(): It removes the lagging white space
string ="@@@@@@@@@Code  @   gnan@@@@@@@@@@"
print(string)
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

```python
# split(): It returns the list of substring and by default splits at white space
string = "The ocean is blue and sky is BLUE 12342@"
print(string.split())
print(string.split('is'))
print(string.split('Z'))
print(string.split('blue'))
print(string.split('ocean'))
```

```python
# Indexing finding Methods
# index(substring) # It returns the first occurance index of string string if exists,otherwisse "value ERROR"
# find(substring) # It returns the first occurance index of Sub_string if exists,otherwisse "-1"
string = "The ocean is blue and sky is blue"
print(string.index("blue"))
print(string.index('is'))
print(string.index('is', 14)) # It starts checking from 14th index
print(string.index('e', 2))
```

```python
 string = "The ocean is blue and sky is blue"
 print(string.find("0"))
 print(string.find('is'))
 print(string.find('is', 14)) # It starts checking from 14th index
 print(string.find('Z'))
```

```python
# replace(old , new): It replace the old_string wth new_string instring
string = "The ocean is blue sky is blue"
print(string.replace('blue','Red'))
print(string.replace('blue','Red', 1))# It replaces only 1 occurence
```

```python
# "sep".Join(iterable): It joins the all iterable elements with seperator
lst = ['The', 'ocean', 'is', 'blue', 'and', 'sky', 'is', 'BLUE']
print("".join(lst))
print(" ".join(lst))
print("_".join(lst))
```

### string check methods
 - islower():It returns True when entire string in lowercase
 - isupper(): It returns True when entire string is in uppercase
 - istitle():It returns True when string is title format
 - isalpha():It returns True when String contains Alphabets(i.e lower and uppercase)
 - isalnum(): It returns True when string either alphabets or digits
 - isdigit(): It returns True when string contains only digits
 - startswith(substring):It returns true when strings start with the substring
 - endswith(substring):It returns True when string endswith the substring

```python
print("Hello".islower())
print("hello".isupper())
print("1234".isdigit())
print("12abc".isalnum())
print("dhanu".isalpha())
print("Dhanush".istitle())
print("Hello World".startswith("Hello"))
print("Hello World".endswith("World"))
```
