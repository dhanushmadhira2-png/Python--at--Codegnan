```python
class Account:
    Bank_name = "dhanush"

    def __init__(self, user_name, balance, password):
        self.user_name = user_name      # Public
        self._balance = balance         # Protected
        self.__password = password      # Private

    def get_balance(self):
        return self._balance

    def withdraw(self, withdraw_amount):
        if self._balance >= withdraw_amount:
            self._balance -= withdraw_amount
            return f"{withdraw_amount} withdrawn successfully. Current balance is {self._balance}"
        else:
            return "Insufficient balance"

    def deposit(self, deposit_amount):
        self._balance += deposit_amount
        return f"{deposit_amount} deposited successfully. Current balance is {self._balance}"

    def __total_users(self):
        return "Total users: 100"


# Create Object
user = Account("dhanu", 5000, "dhanu@123")

# Deposit
print(user.deposit(2000))

# Access balance using getter
print("Balance:", user.get_balance())

# Update protected variable (for learning purpose)
user._balance = 100000

# Withdraw
print(user.withdraw(5000))
```

```python
class Account:
    Bank_name = "Rajesh"

    # Constructor
    def __init__(self, user_name, balance, password):
        self.user_name = user_name        # Public
        self._balance = balance           # Protected
        self.__password = password        # Private

    # Getter method
    def get_balance(self):
        return self._balance

    # Setter method
    def update_password(self, new_password):
        self.__password = new_password
        return "Password updated successfully."

    # Getter method
    def get_password(self):
        return self.__password

    # Withdraw method
    def withdraw(self, withdraw_amount):
        if self._balance >= withdraw_amount:
            self._balance -= withdraw_amount
            return f"{withdraw_amount} withdrawn successfully. Current balance is {self._balance}"
        else:
            return "Insufficient balance."

    # Deposit method
    def deposit(self, deposit_amount):
        self._balance += deposit_amount
        return f"{deposit_amount} deposited successfully. Current balance is {self._balance}"

    # Private method
    def __total_users(self):
        return "Total users: 100"


# Creating object
user = Account("Raja", 5000, "Raja@123")

# Access public variable
print("User Name:", user.user_name)

# Get balance
print("Balance:", user.get_balance())

# Deposit money
print(user.deposit(5000))

# Withdraw money
print(user.withdraw(2000))

# Update password
print(user.update_password("ram@123"))

# Get password
print("Password:", user.get_password())

#access the protected member syntax: obj._attributename
print(user._balance)
#access private member by using name managing syntax:obj._classname_attributename
print(user._Account__total_users())
```
