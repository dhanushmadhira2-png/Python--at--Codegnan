```python
## Count vowels and consonents in string
a = input()
count=0
for i in a:
    if i in "aeiouAEIOU":
        count+=1
print("vowels:",count)
print("consonents:",len(a)-count)
```

```python
#count even and odd numbers in give list of numbers
a = list(map(int,input().split()))
e=o=0
for i in a:
    if i%2==0:
     e+=1
else:
     o+=1
print("even number:s",e)
print("Odd number:s",o)
```

```python
a=list(map(int,input().split()))
even,odd=[],[]
for i in a:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
print('Even numbers',even)

print('Odd numbers',odd)
    
```

```python
##while loop
lst = [1,2,3,4,3,5,3,2]
target=3
a = []
for i in range(len(lst)):
    if i!=target:
        a.append(i)
print(a)
```

```python
#finding the index of the target element
a = list(map(int,input().split()))
target=int(input())
for i in range(len(a)):
    if a[i]==target:
        print(i)
        break
else:
    print(-1)
```
