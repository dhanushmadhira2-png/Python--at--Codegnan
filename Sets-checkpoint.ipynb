#### set methods
- Adding Elements to set
 - add(ele): It adds element to set if element not exists in set, otherwise nothing
 - update(iterable): It adds all iterable elements to set
- remove(ele): It removes the element from setnif element exists in set otherwise `KeyError`
  - discard(ele): It removes the element from set if element exists in set otherwise nothing
- Set Operations
- pop(): It remove and returns the element
 - union(anthor_sets): It returns the all common elements from all sets
 - intersection(anthor_sets): It returns the common elements from the all sets
 - difference(anthor_sets): It returns the set element those are not lies in anthor sets
 - symmetric_difference(anthor_sets): It returns the Non-common elements from all
   

```python
s = {1,2,3,-5}
s.add(4)
print(s)
s.add(1) # 1 doesnt add to set because set contains 1
print(s)
s.remove(1) 
print(s)
#s.remove(1) # KeyError
s.discard(1)
print(s)
```

```python
s1 = {1,2,5,7}
s2 = {3,4,5,6,7}
s3 = {7,9}
print(s1.union(s2))
print(s1.union(s2, s3))
print(s1.intersection(s2))
print(s1.intersection(s2, s3))
print(s1.difference(s2))
print(s1.difference(s2, s3))
print(s1.symmetric_difference(s2))
```

```python
s1 = {1,2,5,7}
s2 = {3,4,5,6,7}
s3 = {7,9}
print(s1 | s2) # / - union
print(s1 | s2 | s3)
print(s1 & s2) # &-intersection
print(s1 & s2 & s3)
print(s1 -s2)# - for Difference
print(s1 - s2 - s3)
print(s1 ^ s2)# for symmetric
print(s1 ^ s2 ^ s3)
```

```python
#This method updates the set object value
s1 = {1,2,3}
s2 = {2,3,4,5}
s1.intersection_update(s2)
print(s1)
s1.difference_update(s2)
print(s1, s2)
s1.difference_update(s2)
print(s1, s2)
s1.symmetric_difference_update(s2)
print(s1)
```
