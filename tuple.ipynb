```python
#access values by using keys
details = {'name':'Ravi', 'age':20, 'course':'python', 'marks': [85,90,86], 'age' :50}
print(details['course'])
print(details['age'])
#updating the dictionary value
details['age'] += 1
details['course'] = 'java'
print(details)
#item assignment
details['location'] = "India" #It adds new item to dictionary
print(details)
```

### Dictionary Methods
- Get value
- get(key, default_val): It returns the value if key exists in dictionary otherwise
 'default_Val'
-Update values and Add items to dictionary
 -Update(new_dictionary): Dictionary Updates with New_dictionary Items
- Remove items from dictionary
- pop(key): It returns value and remove item if key is exists in dictionary other wise 'KeyError'
-popitem(): It returns tuple of key value pair and removes the last item from dictionary
- Get keys, value and items from dictionary
- keys(): It returns list of dictionary keys
-values():It returns list of dictionary values
-items():It list of tuple Key-value oairs

```python
details = {'name':'Ravi', 'age':20, 'course':'python', 'marks': [85,90,86], 'age' :50}
#print(details['location'])
print(details.get('name'))
print(details.get('location'))
print(details.get('age', 18))
print(details.get('location', "Location Key not exists"))
```

```python
details = {'name':'Ravi', 'age':20, 'course':'python'}
d = {'marks':[85,90,86],'age':50}
details.update(d)
print(details)
```

```python
details = {'name':'Ravi', 'age':20,'course':'python', 'marks':[85,90,86], 'age':50}
print(details)
print(details.pop('age'))
print(details)
print(details.popitem())
print(details)
```
