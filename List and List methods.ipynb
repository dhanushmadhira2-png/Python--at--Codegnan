### list methods
- Element Adding Methods
  - append(item):It adds item at end of the list
  - extend(itarable):It adds all iterable elements at end of the list
  -insert(pos, item):This Method adds the item at specified position

```python
lst = [1,2,3,4]
lst.append(10)
print(lst)
lst.append('Code')
print(lst)
lst.append([10,20,30])
print(lst)
lst.extend([100,200,300])
print(lst)
lst.insert(1,2000)
print(lst)
```

```python
lst = [1,2,3,4]
print(lst+[[1,2,3]])
```

## Element remove Methods
 - remov(val): It removes first occurence of value if value exist in list,otherwise it raises 'ValueError'
-pop(): It remove and returns the last of element of list
-pop(pos):It remove and returns the specified position element from list
-clear():It removes all element from list
- del:It Python Keyword,used to delete values

```python
lst = [10,20,30,40]
lst.remove(20)
print(lst)
#lst.remove(100) # valueError: list.remmove(x):x not in list
val = lst.pop()
print(val, lst)
val2 = lst.pop(1) #it removes the lst position element
print(val2, lst)
#print(lst.pop(10))# Index Error: pop index out of range
del lst[0]
print(lst)
lst.clear()
print(lst)
```
