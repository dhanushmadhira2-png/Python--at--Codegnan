```python
# Inheritance the parent class constructor in child class
class Person:
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old and {self.gender}"
# child class
class Student(Person):
    def __init__(self, stdid, name, age,gender):
        self.stdid = stdid
        super().__init__(name, age, gender)
        
    def percentage(self, marks:list):
        return sum(marks)/len(marks)

s = Student(101,"dhanush", 20, "male")
print(s.name)
print(s.details())
print(s.percentage([90,89,88]))
```

```python
# Multilevel inheritance 
class employee:
    Company="Red Bull"
    def __init__(self,eid:int,name:str,salary:int):
        self.eid=eid
        self.name=name
        self.salary=salary
    def calculate_salary(self,bonous=1500):
        return self.salary+bonous
class teamlead(employee):
    def __init__(self,eid:int,name:str,salary:int,project_name:str):
        self.project_name=project_name
        super().__init__(eid,name,salary)
    def project_status(self,status):
        return f"{self.project_name} project status:{status}% completed"
class manager(teamlead):
    def __init__(self,eid:int,name:str,salary:int,project_name:str,no_projects:int):
        self.no_projects=no_projects
        super().__init__(eid,name,salary,project_name)
a=employee(101,"Max",12000)
b=teamlead(105,"Verstappen",455000,"Red Bull")
c=manager(110,"dhanush",2990000,"F1 Tracking ",3)
print(a.calculate_salary())
print(b.project_status(89))
print(b.calculate_salary(2000))
print(c.project_status(85))
print(b.Company)
```

```python
class Father:
    def __init__(self, father_name):
        self.father_name = father_name
    def working1(self):
        return "Father working as  Employee"
class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name
    def working2(self):
        return "Mother is a housewife"
class child(Father, Mother):
    def __init__(self, name, father_name, mother_name):
        self.name = name
        Mother.__init__(self,mother_name)
        Father.__init__(self,father_name)
    def working(self):
        print(super().working1())
        print(super().working2())
c = child("dhanush","siva","suma")
print(c.father_name)
print(c.mother_name)
print(child.__mro__) # Method resolution order
print(c.working())
```
