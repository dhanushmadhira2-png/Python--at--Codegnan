```python
## Method overriding
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def details(self):
        return f"My name is {self.name}"
class student(Person):
    def __init__(self,name,age,rollnum):
        super().__init__(name,age)
        self.rollnum=rollnum
    def details(self):
        return f"My name is {self.name},age is {self.age},rollnum is {self.rollnum}"
a = student("Dhanush",20,21)
b = Person("Max Verstappen",21)
print(a.details())
print(b.details())
```

```python
# Operator Overloading
class Point:
    def __init__(self, x):
        self.x = x
    def __str__(self):# string magic method
        return f" This point class and value is:{self.x}"
    def __add__(self, other):
        return self.x + other.x
    def __sub__(self, other):
        return self.x- other.x
p1 = Point(5)
p2 = Point(3)
print(p1.x + p2.x)
print(p1 + p2)
print(p1-p2)
```
