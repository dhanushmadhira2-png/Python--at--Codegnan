```python
#find the number is prime number or not using function
def prime_check(n:int)->bool:
    for i in range(2,n//2++1):
        if n % i==0:
            return False
        return True
num = int(input())
print(prime_check(num))
        
```

```python
#Types of arguments
#1.Positional arguments: the arguments are assigns to parameters as per the positional order
def sub(x:int,y:int)->int:
    c = x- y
    return c
#function calling block
a, b = map(int,input().split())
res = sub(a,b)
print(res)
```

```python
#Types of arguments
#2. Keyword arguments: the arguments are assigns to parameters as per the positional order
def sub(x:int,y:int)->int:
    c = x- y
    return c
#function calling block
a, b = 10,20
print(sub(x = a,y = b))
print(sub(y = b,x = a))
```

```python
def addition(a:int, b:int, c:int, d:int)->int:
    s = a+b+c+d
    return s
#function calling block
a, b, c, d = 2, 3, 4, 5
print(addition(a,b,c,d))
```

```python
#3.Default Arguments  :The parameters defined with same defalut values,and it assign
#the argument value f pass the values for default parameters otherwise,
#it will assign default values
def add(a,b,c=0, d=0):
    res = a+b+c+d
    return res
print(add(10,20))
print(add(10,20,5))
print(add(10,20,2,1))
print(add(10))
```

```python
def multiply(a,b,c=1,d=1):
    product = a*b*c*d
    return product
print(multiply(4,1))
print(multiply(9,3,1))
print(multiply(10,2,8,7))
```
