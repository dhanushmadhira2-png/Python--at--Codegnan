```python
class Account:
    Bank = "Max Verstappen"

    def __init__(self, username, balance, password):
        self.username = username
        self.balance = balance
        self.password = password

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            return f"₹{amount} withdrawn successfully.\nCurrent Balance: ₹{self.balance}"
        return "Insufficient Balance"

    def deposit(self, amount):
        self.balance += amount
        return f"₹{amount} deposited successfully.\nCurrent Balance: ₹{self.balance}"

# Create object
user = Account("karthik", 5000, "kar@123")

# Deposit
print(user.deposit(1000))

# Change balance
user.balance = 100000

# Withdraw
print(user.withdraw(50000))
        
        
   
    
```
