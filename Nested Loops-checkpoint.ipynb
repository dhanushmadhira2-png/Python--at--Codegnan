```python
##Square shape '*' pattern
n=5
for i in range(n):
    for j in range(n):
        print('*',end=" ")
    print()
```

```python
n=5
for i in range(n):
    for j in range(n):
        print('1',end=" ")
    print()
```

```python
n=4
for i in range(0, n+1):
    for j in range(1, n+1):
        i +=1
        j +=1
        print(i, end=" ")
    print()
```

```python
n = 5
for i in range(n):
    for j in range(n):
        print( n*i+j+1, end="\t")
    print()
```

```python
n = 5
for i in range(n):
    for j in range(n):
        if (i+j)%2==0:
            print(0, end=" ")
        else:
            print(1, end=" ")
    print()
```
